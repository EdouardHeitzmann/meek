# Degree-0 Setting: 

We have a bunch of ballot-comparison outputs, $X_j\in \{-2,-1,0,1,2\}$.

Let's shift and normalize them so they live in $[0,1]$, by doing $Y_j = (X_j+2)/4$. Define $\theta = (1/N)\sum_{j=1}^N Y_j$.

The margin $M_{lc}$ is recorded as $15$; the null is true if $\sum_{j=1}^N X_j < -15$; equivalently, $\theta < \mu_0 = 1/2 - 15/(4\cdot297)$.

```python
import numpy as np

def conditional_mu_deg0(starting_margin, sampling_history, N):
    current_margin = starting_margin + sum(np.array(sampling_history)*4-2)
    print(f"Current margin: {current_margin}")
    remaining_sample_size = N - len(sampling_history)
    assert remaining_sample_size > 0, "Sampling history exceeds total sample size"
    return 1/2 - current_margin / (4 * remaining_sample_size)

def update_Tj(previous_T_j, Y_j, mu_deg0, eta_deg0):
    return previous_T_j *(Y_j * eta_deg0/mu_deg0 + (1 - Y_j) * (1 - eta_deg0)/(1 - mu_deg0))

# Test 1: all X_j are 0

C = [1/2]*20_000 + [1]*20_250 + [0]*19_750
B = [1/2]*20_000 + [1]*20_250 + [0]*19_750
w = [c - b for c, b in zip(C, B)]
Ac = sum(C)/len(C)
v = 2*Ac -1 

x = [(1-w_i)/(2-v) for w_i in w]

#print(v, Ac, sum(C), x[0])

lambda_factor = 1.9
T = [1+ lambda_factor * (x_i-1/2) for x_i in x]
M=1 
alpha = 0.05

for i, T_i in enumerate(T):
    M *= T_i
    print(M)
    if M >= 1/alpha:
        print(f"Reject null hypothesis at step {i+1} with M = {M}")
        break
```
